# 2. String Slicing and Indexing:

# Write a Python program using the above concatenated string as input and performs the following tasks:

# (a) Print the first character of the string.

```python
string = 'Hello zara, welcome to Python programming'
print(string)
print("First character :",string[0])
```

```python
# Python uses positive indexing, which starts from 0. Therefore, string[0] returns the first character, which is H.
```

# (b) Print the last character of the string.

```python
print("Last character :",string[-1])
```

```python
# In Python, negative indexing starts from -1 at the right side of the string. Therefore, string[-1] is used to print the last character of a string.
```

# (c) Print the first 5 characters of the string.

```python
string = 'Hello zara, welcome to Python programming'
print('First 5 characters:', string[0:5])
```

```python
# string stores the given sentence "Hello Zara, welcome to Python programming".
# string[0:5] is used to extract the first five characters from the string. 0 represents starting index and 5 is ending index. 
# String slicing is used to extract a specific part of a string.
```

# (d) Print the last 11 characters of the string.

```python
print("The last 11 characters :",string[-11:])
```

```python
# String[-11:] extracts the last 11 characters from the string.
# -11 is the starting index. 
# Python uses negative indexing to access characters from right to left.
```

# (e) Print the string in reverse.

```python
print("Reversed string:", string[::-1])
```

```python
# string[::-1] is used to reverse the entire string.
# First colon indicates starting index and second colon indicates ending index.
# -1 represents the value, which tells Python to move from right to left.
# print() displays the reversed string.
```

# (f) Use slicing and print the word “Python” from the existing string.

```python
print("Extracted word from the string :",string[22:29])
```

# 3. String Methods:

# Write a Python program that takes a string, strM = “Python beginner tutorial” and perform the following tasks:

# (a) Convert the sentence to uppercase.

```python
strM = "Python beginner tutorial"
print(strM)
print("Uppercase sentence :", strM.upper())
```

```python
# The built-in string method upper() is used to convert all alphabetic characters in a string into uppercase letters.
```

# (b) Convert the sentence to lowercase.

```python
print("Lowercase sentence :",strM.lower())
```

```python
# The built-in string method lower() is used to convert all alphabetic characters in a string into lowercase letters.
```

# (c) Use Capitalize and return the sentence to the original input form.

```python
print("Capitalized sentence :",strM.capitalize())
```

```python
# The capitalize() method is used to convert the first character of a string into uppercase and the remaining characters into lowercase.
```

# (d) Count the total number of occurrences of character ‘t’ in the string.

```python
count = strM.count('t')
print("Total number of occurence of character 't' :",count)
```

```python
# the built-in count() method is used to count the number of occurrences of a specified character or substring in a string.
```

# (e) Replace all occurrences of “Python” with “Machine Learning” in the input string strM = “Python beginner tutorial”

```python
new_string = strM.replace("Python","Machine Learning")
print("Replaced string:", new_string)
```

```python
# The replace() method is used to replace a specific word.
# strM.replace("Python", "Machine Learning") replaces "Python" with "Machine Learning".
# The replaced string is stored in the variable new_string.
# print() displays the new_string.
```

# Create 1st tuple with values -> (10, 20, 30), 2nd tuple with values -> (40, 50, 60):

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)
t_combine = t1 + t2
print("Combined tuple :", t_combine)
```

```python
# t1 = (10, 20, 30) — The first tuple.
# t2 = (40, 50, 60) — The second tuple.
# t1 + t2 — concatenates (joins) the two tuples into one.
# t_combine — stores the newly combined tuple.
# PRINT the t_combine value
```

# (b) Repeat the elements of “t_combine” 3 times

```python
print("Repeated tuple :",t_combine*3)
```

```python
# t_combine * 3 repeats the entire tuple three times.
# The * operator repeats the elements of a tuple the specified number of times.
```

```python
print("The third element : ",t_combine[2])
```

```python
# t_combine contains (10, 20, 30, 40, 50, 60)
# Tuple indexing starts at 0 ie, index[0]= 10
# index[1]=20,index[2]=30 finally we found the 3rd element from the tuple
# Then print the value
```

```python
print("First three elements from t_combine :",t_combine[:3])
```

```python
# t_combine[:3] extracts elements from the beginning up to index 3
# slicing [:3] to access the first three elements of a tuple.
```
