# 1. String Concatenation:

String concatenation is an important concept in Python used to combine multiple strings into one string. It is useful for displaying messages, combining user input, creating sentences, and formatting output.


# Write a Python program that takes two strings i.e string 1 “Hello ”, 
# string 2 get name as input from the user and concatenates them together. Display the concatenated string as the output.

```python
string1 = 'Hello '
string2 = input("Enter your Name: ")
concatenated_string = string1 + string2
print(concatenated_string )
```

# Now concatenate string 3 “, welcome to Python programming” to the existing string and display the output string.


```python
string3 = ', welcome to Python programming'
string4 = concatenated_string + string3
print(string4)
```

```python
# First, the variable string1 is assigned the text 'Hello '. 
# Next, the input() function asks the user to enter their name, and the entered name is stored in the variable string2. 
# Then, the + operator is used to combine string1 and string2. 
# This process is called string concatenation, which means joining two or more strings together.
# The combined result is stored in the variable concatenated_string. 
# Finally, the print() function displays the combined string on the screen.
```

# 2. String Slicing and Indexing:

String indexing and slicing are two important concepts in Python used to access individual characters or extract a part of a string.
•	Indexing is used to access a particular character.
•	Slicing is used to extract a part of a string.


# Write a Python program using the above concatenated string as input and performs the following tasks:

# (a) Print the first character of the string.

```python
string = 'Hello zara, welcome to Python programming'
print(string)
print("First character :",string[0])
```

```python
# Python uses positive indexing, which starts from 0.
# Therefore, string[0] returns the first character, which is H.
```

# (b) Print the last character of the string.

```python
print("Last character :",string[-1])
```

```python
# In Python, negative indexing starts from -1 at the right side of the string. Therefore, string[-1] is used to print the last character of a string.
```

# (c) Print the first 5 characters of the string.

```python
string = 'Hello zara, welcome to Python programming'
print('First 5 characters:', string[0:5])
```

```python
# string stores the given sentence "Hello Zara, welcome to Python programming".
# string[0:5] is used to extract the first five characters from the string. 0 represents starting index and 5 is ending index. 
# String slicing is used to extract a specific part of a string.
```

# (d) Print the last 11 characters of the string.

```python
print("The last 11 characters :",string[-11:])
```

```python
# String[-11:] extracts the last 11 characters from the string.
# -11 is the starting index. 
# Python uses negative indexing to access characters from right to left.
```

```python
# (e) Print the string in reverse.
```

```python
print("Reversed string:", string[::-1])
```

```python
# string[::-1] is used to reverse the entire string.
# First colon indicates starting index and second colon indicates ending index.
# -1 represents the value, which tells Python to move from right to left.
# print() displays the reversed string.
```

# (f) Use slicing and print the word “Python” from the existing string.

```python
print("Extracted word from the string :",string[22:29])
```

```python
# The expression string[22:29] is string slicing, which is used to extract a portion of a string.
# Here, 22 represents the starting index, and 29 represents the ending index.
# The character at index 29 is not included. 
# Therefore, Python extracts the characters from index 22 to index 28 and displays them along with the message "Extracted word from the string :". 
```

# Tuples (Creation, Modification and Access) : 

A tuple in Python is an ordered collection of elements created using parentheses (). It can store different types of values. Tuples support indexing and slicing to access elements. They are immutable, meaning their elements cannot be changed after creation. We can concatenate tuples using + and repeat elements using *.


# Create 1st tuple with values -> (10, 20, 30), 2nd tuple with values -> (40, 50, 60):

# (a) Concatenate the two tuples and store it in “t_combine”

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)
t_combine = t1 + t2
print("Combined tuple :", t_combine)
```

```python
# t1 = (10, 20, 30) — The first tuple.
# t2 = (40, 50, 60) — The second tuple.
# t1 + t2 — concatenates (joins) the two tuples into one.
# t_combine — stores the newly combined tuple.
# PRINT the t_combine value
```

# (b) Repeat the elements of “t_combine” 3 times

```python
print("Repeated tuple :",t_combine*3)
```

```python
# t_combine * 3 repeats the entire tuple three times.
# The * operator repeats the elements of a tuple the specified number of times.
```

# (c) Access the 3rd element from “t_combine”

```python
print("The third element : ",t_combine[2])
```

```python
# t_combine contains (10, 20, 30, 40, 50, 60)
# Tuple indexing starts at 0 ie, index[0]= 10
# index[1]=20,index[2]=30 finally we found the 3rd element from the tuple
# Then print the value
```

# (d) Access the first three elements from “t_combine”

```python
print("First three elements from t_combine :",t_combine[:3])
```

```python
# t_combine[:3] extracts elements from the beginning up to index 3
# slicing [:3] to access the first three elements of a tuple.
```

# (e) Access the last three elements from “t_combine”

```python
print("Last three elements:",t_combine[-3:])
```

```python
# Use negative indexing and slicing to access the last three elements of a tuple.
# index[-1]=60,index[-2]=50,index[-3]=40.These are the last three elements from t_combine
# print the result
```
